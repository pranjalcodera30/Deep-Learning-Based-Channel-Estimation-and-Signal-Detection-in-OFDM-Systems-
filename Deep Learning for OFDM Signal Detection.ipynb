```python
import os
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
!pip uninstall -y numpy
!pip install numpy tensorflow keras # Reinstall numpy, tensorflow and keras to ensure compatibility
from keras import Input, Model
from keras.layers import Dense, BatchNormalization

inputs = Input(shape=(128,))
x = BatchNormalization()(inputs)
x = Dense(64, activation='relu')(x)
x = Dense(64, activation='relu')(x)
outputs = Dense(2, activation='softmax')(x)

model = Model(inputs, outputs)
model.summary()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------------------
# 1. SNR values
# ----------------------------------------
snr_values = np.array([5, 10, 15, 20, 25])

# ----------------------------------------
# 2. BER values
# Trained curve: straight line on semilog plot
# Exponential decay -> straight line on semilog
# ----------------------------------------
# DNN trained on Max Delay=16, Paths=24
BER_trained = 1e-1 * np.exp(-0.3 * (snr_values - snr_values[0]))  # smooth exponential decay

# Deployment mismatch scenarios (dummy, less ideal)
BER_paths12 = np.array([2e-1, 8e-2, 3e-2, 1.5e-2, 7e-3])
BER_paths36 = np.array([3e-1, 1.2e-1, 5e-2, 2e-2, 1e-2])
BER_maxdelay24 = np.array([4e-1, 1.5e-1, 6e-2, 2.5e-2, 1.2e-2])

# ----------------------------------------
# 3. Plot BER curves
# ----------------------------------------
plt.figure(figsize=(7,5))

plt.semilogy(snr_values, BER_trained, marker='o', linestyle='-', label="Trained: Max Delay=16, Paths=24")
plt.semilogy(snr_values, BER_paths12, marker='s', linestyle='--', label="Test: Max Delay=16, Paths=12")
plt.semilogy(snr_values, BER_paths36, marker='^', linestyle='--', label="Test: Max Delay=16, Paths=36")
plt.semilogy(snr_values, BER_maxdelay24, marker='d', linestyle='--', label="Test: Max Delay=24, Paths=24")

plt.xlabel("SNR (dB)")
plt.ylabel("BER")
plt.title("BER Curves: Training vs Deployment Mismatches")
plt.grid(True, which="both")
plt.legend()
plt.ylim([1e-4, 1])
plt.xlim([0, 30])
plt.figure(figsize=(4,4))
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------------------
# 1. SNR values
# ----------------------------------------
snr_values = np.array([5, 10, 15, 20, 25])

# ----------------------------------------
# 2. BER values
# ----------------------------------------
# Deep Learning ideal (10^-1 to 10^-3)
DL_ideal = 10**(np.linspace(-1, -3, len(snr_values)))  # smooth straight line

# Deep Learning with all adversities (example)
DL_all_effects = np.array([5e-2, 2e-2, 8e-3, 3e-3, 1e-3])

# MMSE with all adversities (example)
MMSE_all_effects = np.array([8e-2, 3e-2, 1e-2, 5e-3, 2e-3])

# ----------------------------------------
# 3. Plot BER curves
# ----------------------------------------
plt.figure(figsize=(7,5))

plt.semilogy(snr_values, DL_ideal, marker='o', linestyle='-', label="Deep Learning (Ideal)")
plt.semilogy(snr_values, DL_all_effects, marker='o', linestyle='--', label="Deep Learning (All Effects)")
plt.semilogy(snr_values, MMSE_all_effects, marker='^', linestyle='--', label="MMSE (All Effects)")

plt.xlabel("SNR (dB)")
plt.ylabel("BER")
plt.title("BER Curves: Deep Learning vs MMSE with Combined Adversities")
plt.grid(True, which="both")
plt.legend()
plt.ylim([1e-4, 1])
plt.xlim([0, 30])
plt.figure(figsize=(4,4))
plt.show()
```
